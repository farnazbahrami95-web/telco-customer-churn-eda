```python
def ecdf(x):
    x = np.sort(x[~np.isnan(x)])
    y = np.arange(1, len(x)+1 / len(x))
    return x, y
```

```python
for col in num_cols:
    x, y = ecdf(df[col].values)
    plt.plot(x, y, marker='.', linestyle='none')
    plt.title(f'ECDF: {col}')
    plt.xlabel(col)
    plt.ylabel('ECDF')
    plt.show()
```
